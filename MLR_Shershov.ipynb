2. Проиллюстрировать качество полученной модели и выписать её уравнение.


```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_percentage_error
```

```python
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Размер набора данных:", df.shape)
display(df.head())
```

```python
df["Extracurricular Activities"] = df["Extracurricular Activities"].map({
    "Yes": 1,
    "No": 0
})

X = df[
    [
        "Hours Studied",
        "Previous Scores",
        "Extracurricular Activities",
        "Sleep Hours",
        "Sample Question Papers Practiced"
    ]
]

y = df["Performance Index"]

print("Количество объектов:", len(df))
print("Количество признаков:", X.shape[1])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,
    random_state=42
)

print("Обучающая выборка:", X_train.shape)
print("Тестовая выборка:", X_test.shape)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
coefficients = pd.DataFrame({
    "Признак": X.columns,
    "Коэффициент": model.coef_
})

display(coefficients)

print("Свободный член:", round(model.intercept_, 4))

equation = f"Performance Index = {model.intercept_:.4f}"

for name, coef in zip(X.columns, model.coef_):
    sign = "+" if coef >= 0 else "-"
    equation += f" {sign} {abs(coef):.4f} * ({name})"

print("\nУравнение регрессии:")
print(equation)
```

```python
y_pred = model.predict(X_test)

result = pd.DataFrame({
    "Реальное значение": y_test.to_numpy(),
    "Предсказанное значение": y_pred
})

result["Абсолютная ошибка"] = (
    result["Реальное значение"] - result["Предсказанное значение"]
).abs()

display(result.head(10))
```

```python
r2 = r2_score(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred) * 100

print(f"Коэффициент детерминации R²: {r2:.5f}")
print(f"Средняя ошибка аппроксимации MAPE: {mape:.2f}%")
```

```python
print("РЕЗУЛЬТАТЫ РАБОТЫ")
print(f"Количество объектов: {len(df)}")
print(f"Количество признаков: {X.shape[1]}")
print(f"Обучающая выборка: {len(X_train)} объектов")
print(f"Тестовая выборка: {len(X_test)} объектов")
print(f"R² = {r2:.5f}")
print(f"MAPE = {mape:.2f}%")
print("\nПолученное уравнение регрессии:")
print(equation)

if r2 >= 0:
    print(
        f"Модель объясняет {r2 * 100:.2f}% вариации показателя Performance Index "
        "на тестовой выборке."
    )
else:
    print(
        "Значение R² отрицательное: на тестовой выборке модель работает "
        "хуже прогноза средним значением целевой переменной."
    )

print(f"Средняя ошибка прогноза составляет {mape:.2f}%.")
```
